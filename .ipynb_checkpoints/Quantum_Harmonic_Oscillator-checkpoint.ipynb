# Quantum Harmonic Oscillator

The simple harmonic oscillator is ubiquitous in physics, applied mathematics and engineering. 
This is because any potential $V(x)$ can be approximated by a harmonic potential in the neighbourhood of a local minima $x_0$. This is evident using the Taylor expansion of $V(x)$ around $x_0$: $$V(x) = V(x_0) + V'(x_0)(x - x_0) + \frac{1}{2}V''(x_0)(x - x_0)^2 + ... \approx \frac{1}{2}k(x - x_0)^2$$ where $k = V''(x_0)$ is the (effective) spring constant. This classical model does not generalise to systems where quantum effects are significant, such as the vibrations of diatomic molecules and phonons in crystalline solids. Hence, a quantum mechanical treatment of the harmonic oscillator is required. 

Quantum systems evolve according to the Schrodinger equation: $$\hat{H}\psi(x,t) = i\hbar\frac{\partial \psi(x,t)}{\partial t}$$ with the Hamiltonian $\hat{H} = \hat{T} + \hat{V}$ where $\hat{T} = \frac{-\hbar^2}{2m} \frac{\partial^2}{\partial x^2}$ is the kinetic energy operator and $\hat{V}$ is the system potential. 

The quantum formulation of the harmonic oscillator requires that the system potential is $\hat{V}(x) = \frac{1}{2} m\omega^2 x^2$, where the classical expression for angular frequency $\omega = \sqrt{k/m}$ is used in place of the spring constant $k$. Such a potential permits the Schrodinger equation to have separable solutions.
Hence, consider the time-independent Schrodinger equation: $$ \frac{-\hbar^2}{2m} \frac{\partial^2 \psi(x)}{\partial x^2} + \frac{1}{2} m\omega^2 x^2 \psi(x) = E\psi(x)$$

This eigenvalue equation has known analytical solutions given by: $$\psi_n(x) = A_nH_n(\beta x)e^{\frac{-\beta x^2}{2}}$$ with corresponding eigenvalues $$E_n = (n + \frac{1}{2})\hbar \omega$$ where $\beta = \sqrt{m\omega/\hbar}$ , $A_n$ is the normalisation constant and $H_n$ is the Hermite polynomial of degree $n$. 

This project solves the TISE for a quantum harmonic oscillator system using the central differences numerical method for approximating first and second derivatives. The accuracy and limitations of this method are investigated by considering the relative error in the numerically determined $E_n$. The eigenstates calculated are subsequently used to construct a Gaussian wavepacket in the harmonic potential, which refines the initial understanding of the limitations of the central differences method. 

## 1.0) Numerical Solution of TISE for Quantum Harmonic Oscillator

Nunerical solutions to the TISE (an eigenvalue equation) are found by approximating the Hamiltonian $\hat{H}$ as an $N \times N$ matrix, where $N = 4000$, by discretising the domain and then determining the eigenstates and eigenvalues of this matrix. The domain, which extends from $x = -20$ to $x = +20$, is divided into $N$ grid points, labelled $x_i$ for $ i = 0, 1,...,N-1$, with length $dx = 0.01$ between each pair of grid points. 

The central differences method, from the finite differences family of numerical methods for approximating derivatives, is employed to approximate the second derivative in $\hat{T}$ at each grid point. For a general function $f$ over a grid of discrete points $y_j$ with grid spacing $h$, the second derivative $f''$ at $y_j$ is given by: 
$$ f''(y_j) \approx \frac{f(y_j + h) - 2f(y_j) + f(y_j - h)}{(h)^2}$$
When applied to the second derivative in $\hat{T}$ and combined with $\hat{V}$, a diagonal matrix with $\frac{1}{2} m\omega x_i^2$ on the leading diagonal, the Hamiltonian is approximated by: 
$$\hat{H} \approx \begin{matrix}
                        1 & 2 \\
                        3 & 4
                  \end{matrix}$$

The finite differences family of numerical derivative methods was selected for this task as the simplicity of implementing any one of them using matrices is ideal for taking advantage of the eigenvalue structure of the equation. The central difference method is a $\mathcal{O}(h^2)$ method, meaning that the leading truncation error is proportional to $h^2$. This contrasts the forwards and backwards difference methods which are $\mathcal{O}(h)$ methods, so the leading truncation errors is proportional to $h$. For small $h$, $h^2 << h$, so the central difference method provides a more accurate approximation of $\hat{H}$ for $dx = 0.01$ than the forwards and backwards difference methods. Given that there was also little increase in the runtime when implementing the central difference method compared to lower order methods, it was appropriate to employ the central difference method in the approximation of $\hat{H}$.   

The choice of $dx = 0.01$ is critical in understanding the magnitude of both the truncation and roundoff errors in the approximation, as well as impacting the program runtime. This is explored in **1.1) Error Analysis: Truncation vs Roundoff error**.

The combination of $N = 4000$ and $x_{min} = -20 / x_{max} = 20$ for a fixed $dx$ determines the accuracy of higher energy eigenstates, hence the subsequent behaviour of the Gaussian wavepacket in **2.0) Time Evolution of Gaussian Wavepacket in Harmonic Potential**. This is considered in **2.1) Error analysis: Domain Turning point error**.

```python
# define dimensionless parameters

hbar = 1 # natural units: set hbar = 1 
w = 1 # angular frequency
m = 1 # particle mass
```

```python
# discretise position

dx = 0.01 # dx = separation of grid points in 1D domain
# chosen value justified in Error Analysis: Truncation vs Roundoff error

N = 4000 # N = number of grid points in domain
L = N*dx # chosen value justified in Error Analysis: Domain turning point error
```

```python
# function to calculate system Hamiltonian with N and dx as input variable

# Defining the Hamiltonian using a function this way simplifies computation in Error Analysis: Domain turning point error when N and dx become variables.

def Hamiltonian(N,dx):
    L = N*dx # length of domain
    x = np.linspace(-L/2, L/2, N) # grid point values

    # kinetic energy operator T
    T1 = np.diag(np.full(N,-2))+np.diag(np.ones(N-1),1)+np.diag(np.ones(N-1),-1) # central differences method implementation
    T = -(hbar**2 / (2*m*(dx**2)))*T1

    # potential energy operator for V(x) = mw^2 x^2 / 2 harmonic potential
    V1 = np.diag(x**2)
    V = 0.5*m*(w**2)*V1

    # Calculate system Hamiltonian H = T + V
    H = T + V
    return(H)
```

For a continuous spatial domain, the normalisation condition requires $\int |\psi|^2\,dx = 1$. After discretising the domain, this condition is approximated as $\sum_{i} |\psi_i|^2 dx = 1$, where $\psi_i$ are the eigenstates of the approximated Hamiltonian. The eigenstates returned by `eigh` satisfy $\sum_i |\psi_i|^2 = 1$, so each eigenstate differs from the correct normalisation for this method by a factor of $\sqrt{dx}$. This must be corrected for. 

```python
# Calculate eigenvalues and corresponding eigenvectors of Hamiltonian
E, psi = eigh(Hamiltonian(N,dx))

psi = psi/np.sqrt(dx) # renormalise eigenvectors
prob = psi**2 # probability density
```

```python
# Plot wavefunctions vs position
x = np.linspace(-L/2, L/2, N)

fig = plt.figure(figsize=(10,6))
ax = fig.add_subplot(1,2,1)
ax.plot(x, psi[:,0], label=r"$\psi_0$")
ax.plot(x, psi[:,1]+2, label=r"$\psi_1$")
ax.plot(x, psi[:,2]+4, label=r"$\psi_2$")
ax.plot(x, psi[:,3]+6, label=r"$\psi_3$")
ax.plot(x, psi[:,4]+8, label=r"$\psi_4$")
ax.set_xlabel('x')
ax.set_yticks([])
ax.set_title(r"Numerical $\psi(x)$")
plt.legend()

# Plot probability densities vs position
ax = fig.add_subplot(1,2,2)
ax.plot(x, prob[:,0], label=r"$|\psi_0|^2$")
ax.plot(x, prob[:,1]+1, label=r"$|\psi_1|^2$")
ax.plot(x, prob[:,2]+2, label=r"$|\psi_2|^2$")
ax.plot(x, prob[:,3]+3, label=r"$|\psi_3|^2$")
ax.plot(x, prob[:,4]+4, label=r"$|\psi_4|^2$")
ax.set_xlabel('x')
ax.set_yticks([])
ax.set_title(r"Numerical $|\psi(x)|^2$")
plt.legend()
plt.show()
```

First, consider the lowest energy states.

Evidently, for low energy states, the numerical eigenvalues are accurate to 3 significant figures. The error in $E_{n, \text{num}}$ is of the order $10^{-4}$. Since the central differences method is a second order method and for this approximation $dx = 10^{-2}$, which means $(dx)^2 = 10^{-4}$. Hence, the major source of error in the low energy states is the truncation error inherent in the numerical method used. 

To understand how the error in $E_{n, \text{num}}$ evolves as $n$ increases, consider the relative error in $E_{n, \text{num}}$ for all $n$ in the figures below.

```python
# plot relative error in E_n vs n
fig = plt.figure(figsize=(12,6))
ax1 = fig.add_subplot(1,2,1)
ax1.plot(n, E_error, label = 'Relative error')
ax1.axhline(0.0001, color='k', linestyle='--')
#ax1.axvline(200, 0.01, 0.8)
ax1.set_xlabel(r'n')
ax1.set_ylabel(r'Relative error in $E_n$')
plt.legend()

# plot analytical and numerical energies vs n
ax2 = fig.add_subplot(1,2,2)
ax2.plot(n, E_analytical*10**(-3), label=r"$E_{n, \text{an}}$")
ax2.plot(n, E*10**(-3), label=r"$E_{n, \text{num}}$")
ax2.set_xlabel(r'$n$')
ax2.set_ylabel(r'$E \times 10^{-3}$')
plt.legend()
plt.show()
```

This figure indicates that for low energy eigenstates, up to $n \approx 200$, the relative error remains on the order of $10^{-4}$, reiterating that the truncation error inherent in the central differences method is the major source of error. However, for higher energy eigenstates, a secondary source of error becomes significant and surpasses the magnitude of the truncation error, demonstrated in the left hand plot by the rapid increase in the relative error from $n \approx 200$, and in the right hand plot as $E_{n, \text{num}}$ diverges rapidly from $E_{n, \text{an}}$ for the same values of $n$. 

The following sections investigates 2 possibilities for this secondary source of error; firstly the roundoff error, which, similar to the truncation error, is an inherent property of the numerical method, and secondly the error due to the domain width $L$, which arises from approximating the derivative at the boundaries of the domain. 

### 1a) Error Analysis: Truncation vs Roundoff error

- What is truncation error - dominates at large $dx$
- What is Roundoff error - dominates at low $dx$
- Investigate how total error for this method varies with $dx$ - use sin$(x)$ as example function (truncation inherent to method not function)
- Minimise total error - find turning point between where truncation and roundoff error dominate

```python
# numerically sweep dx over 6 orders of magnitude (via N and L)

L_err = 4*np.pi
N_err = np.logspace(1,7, 200)
```

```python
rel_err = np.zeros(200)

# for each value of dx, find the relative error of the second derivative at x = pi/4 # 

for i in range(200):

    # sweeping dx by fixing L and changing N
    N_i = int(N_err[i])
    dx_err = L_err/N_i
    
    x_err = np.linspace(-L_err/2, L_err/2, N_i)
    s = np.sin(x_err)

    # finite differences central difference method for second derivative
    num_2der = np.zeros(N_i)
    num_2der[1:-1] = (s[2:] + s[:-2] - 2*s[1:-1])/dx_err**2

    # k - closest x value to pi/4
    k = np.argmin(np.abs(x_err - np.pi/4))
    rel_err[i] = np.abs( - np.sin(x_err[k]) - num_2der[k])/np.sin(x_err[k]) # relative error for each dx value
```

```python
# plot relative error vs dx
dx_err = L_err/N_err

fig = plt.figure(figsize=(10,6))
ax = fig.add_subplot(1,1,1)
ax.plot(dx_err, rel_err)
#ax.axvline(dx_err[163])
ax.set_xlabel('dx')
ax.set_ylabel('relative error')
ax.set_xscale('log')
ax.set_yscale('log')
plt.savefig('trun_round_err')
plt.show()
```

Locating turning points:
- separate straight line fits to data for $dx$ < $dx_{min}$ (where roundoff error dominates) and for $dx$ > $dx_{min}$  (where truncation error dominates)
- find intersection = turning point
- using least squares method ***

```python
# function for straight line fits
def straight_line(x,m,c):
    y = m*x + c
    return(y)

# fit straight line parameters for dx > dx_min
popt, pcov = curve_fit(straight_line, np.log10(dx_err[20:130]), np.log10(rel_err[20:130]))
m_r = popt[0] 
c_r = popt[1]

# fit straight line parameters for dx < dx_min
popt1, pcov1 = curve_fit(straight_line, np.log10(dx_err[155:199]), np.log10(rel_err[155:199]))
m_l = popt1[0]
c_l = popt1[1] 
```

```python
# plot relative error vs dx with straight line fits

fig = plt.figure(figsize=(10,6))
ax = fig.add_subplot(1,1,1)
ax.plot(np.log10(dx_err), np.log10(rel_err))
ax.plot(np.log10(dx_err), m_r*np.log10(dx_err) + c_r)
ax.plot(np.log10(dx_err[130:199]), m_l*np.log10(dx_err[130:199]) + c_l)
ax.set_xlabel(r'log_10(dx)')
ax.set_ylabel(r'log_10(relative error)')
plt.show()
```

```python
# turning point for optimising round off vs truncation error for this numerical method

p = np.argmin(np.abs((m_r*np.log10(dx_err) + c_r) - (m_l*np.log10(dx_err) + c_l))) # index of dx_min
print(f"The optimal grid spacing for minimising truncation and round-off error when implementing the central differences method for a second derivative is dx = {dx_err[p]:.8f}.")
```

While minimising both roundoff and truncation error for this numerical method requires $\Delta x \approx 10^{-5}$, a value of $\Delta x = 10^{-2}$ is used. This is because $\Delta x < 10^{-2}$ makes the programs take too long to run.
- Truncation error dominates (roundoff error not significant in comparison ~$10^{-12}$)
- Second order method $\rightarrow$ truncation varies as $O(\Delta x)^2$
- Hence eigenvalues have errors of order greater than/equal to $(\Delta x)^2 = 10^{-4}$ for first few eigenstates

# 2.0) Time Evolution of Gaussian Wavepacket in Harmonic Potential

- Gaussian wavepacket - equation
- Describe known behaviour in harmonic potential - no dispersion etc
- Explain what this section shows - calculates wavepacket and time evolution, justifies use of first 80 eigenstates only, explains errors incurred when using all eigenstates and subsequent turning point error.
- Known solutions of TDSE

```python
# define parameters for intitial position, initial momentum and spread of wavepacket

x_0 = 1
k_0 = 10
sigma = 1
```

```python
# function for initial Gaussian wavepacket

def gaussian_wavepacket(x):
    return((1/(2*np.pi*sigma**2)**0.25)*np.exp(-((x-x_0)**2)/4*sigma**2)*np.exp(1j*k_0*(x-x_0)))
```

```python
# gaussian wavepacket at time t = 0

Psi_initial = np.zeros(N, dtype=complex)
for k in range(N):
    Psi_initial[k] = gaussian_wavepacket(x[k])
```

```python
# plot initial Gaussian wavepacket and probability density
fig = plt.figure(figsize=(10,6))
ax = fig.add_subplot(1,2,1)
ax.plot(x, Psi_initial.real)
ax.set_ylabel('psi')
ax.set_xlabel('x')

ax1 = fig.add_subplot(1,2,2)
ax1.plot(x, np.abs(Psi_initial)**2)
ax1.set_ylabel('psi squared')
ax1.set_xlabel('x')
plt.show()
```

```python
# calculating coefficients c_n to decompose wavepacket into system eigenstates 

n_max = 80 # max number of eigenstates used - minimise error due to L

c_n = np.zeros(n_max, dtype=complex)

# calculate coefficients
for b in range(n_max):
    for a in range(N):
        c_n[b] += psi[a,b]*Psi_initial[a]*dx
```

```python
print(sum(np.abs(c_n)**2))

n = np.arange(n_max)

fig = plt.figure(figsize=(10,6))
ax = fig.add_subplot(1,1,1)
ax.plot(n, np.abs(c_n)**2)
ax.set_xlabel('n')
ax.set_ylabel('c_n^2')
plt.show()
```

```python
# discretise time step t 
M = 201
t = np.linspace(0,20,M)

# initialise Psi_xt
Psi_xt = np.zeros((N,M), dtype=complex)

# use matrix multiplication to find time evolution from Psi_initial
for k in range(M):
    c_nk = c_n*np.exp(-1j*E[:n_max]*t[k]/hbar)
    Psi_xt[:,k] = np.matmul(psi[:,:n_max], c_nk)
```

```python
fig = plt.figure(figsize=(10,6))
ax = fig.add_subplot(2,2,1)
ax.plot(x, Psi_xt[:,0].real)
ax.set_ylabel('psi @ t = 0')
ax.set_xlabel('x')

ax1 = fig.add_subplot(2,2,2)
ax1.plot(x, Psi_xt[:,25].real)
ax1.set_ylabel('psi @ t = 2.5')
ax1.set_xlabel('x')

ax2 = fig.add_subplot(2,2,3)
ax2.plot(x, Psi_xt[:,50].real)
ax2.set_ylabel('psi @ t = 5')
ax2.set_xlabel('x')

ax3 = fig.add_subplot(2,2,4)
ax3.plot(x, Psi_xt[:,75].real)
ax3.set_ylabel('psi @ t = 7.5')
ax3.set_xlabel('x')

plt.show()
```

```python
fig = plt.figure(figsize=(10,6))
ax = fig.add_subplot(2,2,1)
ax.plot(x, Psi_xt[:,10].real)
ax.set_ylabel('psi @ t = 10')
ax.set_xlabel('x')

ax1 = fig.add_subplot(2,2,2)
ax1.plot(x, Psi_xt[:,125].real)
ax1.set_ylabel('psi @ t = 12.5')
ax1.set_xlabel('x')

ax2 = fig.add_subplot(2,2,3)
ax2.plot(x, Psi_xt[:,150].real)
ax2.set_ylabel('psi @ t = 15')
ax2.set_xlabel('x')

ax3 = fig.add_subplot(2,2,4)
ax3.plot(x, Psi_xt[:,175].real)
ax3.set_ylabel('psi @ t = 17.5')
ax3.set_xlabel('x')

plt.show()
```

```python
# vary L via N: L = N*dx
N_tp = np.linspace(100,4000, 51).astype(int) 

# exact eigenvalues for first 100 eigenstates
n = np.arange(100)
E_analytical = (n+0.5)*hbar*w

# stores relative error for 1st 100 eigenstates for values of N
E_rel_err = np.zeros((100,len(N_tp)))
```

## 2.2) Position: expectation and uncertainty

The following determines the expectation values and uncertainties in the position and momentum for the Gaussian wavepacket in a harmonic potential, and confirms that the Heinsenburg Uncertainty Principle is satisfied for all time.

- The uncertainty in observable $a$ with corresponding operator $\hat{A}$ for wavefunction $\psi$ is $\Delta a = \sqrt{\langle a^2 \rangle - \langle a \rangle^2}$.
- The expectation value of $a$ is $\langle a \rangle = \langle \psi | \hat{A} | \psi \rangle$.
- The expectation value of $a^2$ is $\langle a^2 \rangle = \langle \psi | \hat{A}^2 | \psi \rangle$
- In position basis, $\hat{X} = x$

```python
# <x> as function of t
x_exp_t = np.zeros(M)

for j in range(M):
    x_exp_t[j] = sum((np.abs(Psi_xt[:,j])**2)*x*dx)

# <x^2> as function of t
x2_exp_t = np.zeros(M)

for j in range(M):
    x2_exp_t[j] = sum((np.abs(Psi_xt[:,j])**2)*(x**2)*dx)

# uncertainty in x as function of t
x_unc_t = np.sqrt(x2_exp_t - x_exp_t**2)
```

```python
# plots for <x>, <x^2> and delta x respectively as functions of time
fig = plt.figure(figsize=(10,6))

ax1 = fig.add_subplot(3,1,1)
ax1.plot(t, x_exp_t)
ax1.set_xlabel('t')
ax1.set_ylabel('expectation value of x')

ax2 = fig.add_subplot(3,1,2)
ax2.plot(t, x2_exp_t)
ax2.set_xlabel('t')
ax2.set_ylabel('expectation value of x^2')

ax3 = fig.add_subplot(3,1,3)
ax3.plot(t, x_unc_t)
ax3.set_xlabel('t')
ax3.set_ylabel('delta x')
plt.show()
```

# 2.3) Momentum: Expectation and Uncertainty

- In position basis, the momentum operator is $\hat{P} = -i \hbar \frac{d}{dx}$ and $\hat{P}^2 = - \hbar^2 \frac{d^2}{dx^2}$.
- To approximate the first derivative for $\hat{P}$, the central differences method is used as before. This remains a second order numerical method.
- The second derivative for $\hat{P}^2$ is approximated using the same central differences approach as for the original Hamiltonian

```python
ddx = np.diag(np.ones(N-1),1)+np.diag(-np.ones(N-1),-1) # matrix for first x derivative

# <p> as function of t
p_exp_t = np.zeros(M, dtype=complex)
for m in range(M):
    p_exp_t[m] = (-1j*hbar/2)*np.matmul(np.conj(Psi_xt[:,m]), np.matmul(ddx, Psi_xt[:,m]))

# <p^2> as function of t
p2_exp_t = np.zeros(M, dtype=complex)
for n in range(M):
    p2_exp_t[n] = (-1/dx)*np.matmul(np.conj(Psi_xt[:,n]), np.matmul(T1, Psi_xt[:,n])) # T1 is the matrix for the second x derivative as defined in the Hamiltonian function

# uncertainty in p as function of t
p_unc_t = np.sqrt(p2_exp_t - p_exp_t**2)
```

```python
# plots for <p>, <p^2> and delta p respectively as functions of time
fig = plt.figure(figsize=(10,6))

ax1 = fig.add_subplot(3,1,1)
ax1.plot(t, p_exp_t.real)
ax1.set_xlabel('t')
ax1.set_ylabel('expectation value of p')

ax2 = fig.add_subplot(3,1,2)
ax2.plot(t, p2_exp_t)
ax2.set_xlabel('t')
ax2.set_ylabel('expectation value of x^2')

ax3 = fig.add_subplot(3,1,3)
ax3.plot(t, p_unc_t)
ax3.set_xlabel('t')
ax3.set_ylabel('delta p')
plt.show()
```

- The uncertainty principle 

```python
 # check satisfies uncertainty principle

fig = plt.figure(figsize=(10,6))
ax = fig.add_subplot(1,1,1)
ax.plot(t, x_unc_t*p_unc_t, color='r',label='delta x delta p')
ax.hlines(hbar/2, t[0], t[200], label='hbar/2')
ax.set_xlabel('t')
ax.set_ylabel('delta x delta p')
plt.legend()
plt.show()
```
