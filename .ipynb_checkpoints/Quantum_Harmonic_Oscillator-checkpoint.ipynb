```python
import numpy as np
from scipy.linalg import eigh
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from scipy.optimize import curve_fit
from scipy import special
from scipy.signal import find_peaks
```

# Quantum Harmonic Oscillator

The simple harmonic oscillator is ubiquitous in physics, applied mathematics and engineering. 
This is because any potential $V(x)$ can be approximated by a harmonic potential in the neighbourhood of a local minima $x_0$. This is evident using the Taylor expansion of $V(x)$ around $x_0$: $$V(x) = V(x_0) + V'(x_0)(x - x_0) + \frac{1}{2}V''(x_0)(x - x_0)^2 + ... \approx \frac{1}{2}k(x - x_0)^2$$ where $k = V''(x_0)$ is the (effective) spring constant. This classical model does not generalise to systems where quantum effects are significant, such as the vibrations of diatomic molecules and phonons in crystalline solids. Hence, a quantum mechanical treatment of the harmonic oscillator is required. 

Quantum systems evolve according to the Schrodinger equation: $$\hat{H}\psi(x,t) = i\hbar\frac{\partial \psi(x,t)}{\partial t}$$ with the Hamiltonian $\hat{H} = \hat{T} + \hat{V}$ where $\hat{T} = \frac{-\hbar^2}{2m} \frac{\partial^2}{\partial x^2}$ is the kinetic energy operator and $\hat{V}$ is the system potential. 

The quantum formulation of the harmonic oscillator requires that the system potential is $\hat{V}(x) = \frac{1}{2} m\omega^2 x^2$, where the classical expression for angular frequency $\omega = \sqrt{k/m}$ is used in place of the spring constant $k$. Such a potential permits the Schrodinger equation to have separable solutions.
Hence, consider the time-independent Schrodinger equation: $$ \frac{-\hbar^2}{2m} \frac{\partial^2 \psi(x)}{\partial x^2} + \frac{1}{2} m\omega^2 x^2 \psi(x) = E\psi(x)$$

This eigenvalue equation has known analytical solutions given by: $$\psi_n(x) = A_nH_n(\beta x)e^{\frac{-\beta x^2}{2}}$$ with corresponding eigenvalues $$E_n = (n + \frac{1}{2})\hbar \omega$$ where $\beta = \sqrt{m\omega/\hbar}$ , $A_n$ is the normalisation constant and $H_n$ is the Hermite polynomial of degree $n$. 

This project solves the TISE for a quantum harmonic oscillator system using the central differences numerical method for approximating first and second derivatives. The accuracy and limitations of this method are investigated by considering the relative error in the numerically determined $E_n$. The eigenstates calculated are subsequently used to construct a Gaussian wavepacket in the harmonic potential, which refines the initial understanding of the limitations of the central differences method. 

## 1.0) Numerical Solution of TISE for Quantum Harmonic Oscillator

Nunerical solutions to the TISE (an eigenvalue equation) are found by approximating the Hamiltonian $\hat{H}$ as an $N \times N$ matrix, where $N = 4000$, by discretising the domain. This converts the equation into a system of linear equations, which can then be solved by determining the eigenstates and eigenvalues of this matrix. The domain, which extends from $x = -20$ to $x = +20$, is divided into $N$ grid points, labelled $x_i$ for $ i = 0, 1,...,N-1$, with length $dx = 0.01$ between each pair of grid points. 

The central differences method, from the finite differences family of numerical methods for approximating derivatives, is employed to approximate the second derivative in $\hat{T}$ at each grid point. For a general function $f$ over a grid of discrete points $y_j$ with grid spacing $h$, the second derivative $f''$ at $y_j$ is given by: 
$$ f''(y_j) \approx \frac{f(y_j + h) - 2f(y_j) + f(y_j - h)}{(h)^2}$$
When applied to the second derivative in $\hat{T}$ and combined with $\hat{V}$, a diagonal matrix with $\frac{1}{2} m\omega^2 x_i^2$ on the leading diagonal, the Hamiltonian is approximated by: 
$$\hat{H} \approx
\begin{pmatrix} 
    \frac{\hbar^2}{m} + \frac{1}{2}m \omega^2 x_0^2 & -1 & 0 & \cdots & 0 \\
    -1 & \frac{\hbar^2}{m} + \frac{1}{2}m \omega^2 x_1^2 & -1 &\cdots & 0 \\
    \vdots & \vdots & \ddots & \vdots & \vdots \\
    0 & 0 & \cdots & -1 & \frac{\hbar^2}{m} + \frac{1}{2}m \omega^2 x_{N-1}^2 
\end{pmatrix}$$

The finite differences family of numerical derivative methods was selected for this task as the simplicity of implementing any one of them using matrices is ideal for taking advantage of the eigenvalue structure of the equation. The central difference method is a $\mathcal{O}(h^2)$ method, meaning that the leading truncation error is proportional to $h^2$. This contrasts the forwards and backwards difference methods which are $\mathcal{O}(h)$ methods, so the leading truncation errors is proportional to $h$. For small $h$, $h^2 << h$, so the central difference method provides a more accurate approximation of $\hat{H}$ for $dx = 0.01$ than the forwards and backwards difference methods. Given that there was also little increase in the runtime when implementing the central difference method compared to lower order methods, it was appropriate to employ the central difference method in the approximation of $\hat{H}$.   

The choice of $dx = 0.01$ is critical in understanding the magnitude of both the truncation and roundoff errors in the approximation, as well as impacting the program runtime. This is explored in **1.1) Error Analysis: Truncation vs Roundoff error**.

The combination of $N = 4000$ and $x_{min} = -20 / x_{max} = 20$ for a fixed $dx$ determines the accuracy of higher energy eigenstates, hence the subsequent behaviour of the Gaussian wavepacket in **2.0) Time Evolution of Gaussian Wavepacket in Harmonic Potential**. This is considered in **1.2) Error analysis: Domain Turning point error**.

```python
# define dimensionless parameters

hbar = 1 # natural units: set hbar = 1 
w = 1 # angular frequency
m = 1 # particle mass
```

```python
# discretise position

dx = 0.01 # dx = separation of grid points in 1D domain
# chosen value justified in Error Analysis: Truncation vs Roundoff error

N = 4000 # N = number of grid points in domain
L = N*dx # chosen value justified in Error Analysis: Domain turning point error
```

```python
# function to calculate system Hamiltonian with N and dx as input variable

# Defining the Hamiltonian using a function this way simplifies computation in Error Analysis: Domain turning point error when N and dx become variables.

def Hamiltonian(N,dx):
    L = N*dx # length of domain
    x = np.linspace(-L/2, L/2, N) # grid point values

    # kinetic energy operator T
    T1 = np.diag(np.full(N,-2))+np.diag(np.ones(N-1),1)+np.diag(np.ones(N-1),-1) # central differences method implementation
    T = -(hbar**2 / (2*m*(dx**2)))*T1

    # potential energy operator for V(x) = mw^2 x^2 / 2 harmonic potential
    V1 = np.diag(x**2)
    V = 0.5*m*(w**2)*V1

    # Calculate system Hamiltonian H = T + V
    H = T + V
    return(H)
```

For a continuous spatial domain, the normalisation condition requires $\int |\psi|^2\,dx = 1$. After discretising the domain, this condition is approximated as $\sum_{i} |\psi_i|^2 dx = 1$, where $\psi_i$ are the eigenstates of the approximated Hamiltonian. The eigenstates returned by `eigh` satisfy $\sum_i |\psi_i|^2 = 1$, so each eigenstate differs from the correct normalisation for this method by a factor of $\sqrt{dx}$. This must be corrected for. 

```python
# Calculate eigenvalues and corresponding eigenvectors of Hamiltonian
E, psi = eigh(Hamiltonian(N,dx))

psi = psi/np.sqrt(dx) # renormalise eigenvectors
prob = psi**2 # probability density
```

```python
# Plot wavefunctions vs position
x = np.linspace(-L/2, L/2, N)

fig = plt.figure(figsize=(14,6))
ax = fig.add_subplot(1,2,1)
ax.plot(x, x**2/2, color='silver', linestyle='--')
ax.plot(x, psi[:,0]+E[0], color='darkblue', label=r"$\psi_0$")
ax.plot(x, psi[:,1]+E[1], color='blue', label=r"$\psi_1$")
ax.plot(x, psi[:,2]+E[2], color='royalblue', label=r"$\psi_2$")
ax.plot(x, psi[:,3]+E[3], color='dodgerblue', label=r"$\psi_3$")
ax.plot(x, psi[:,4]+E[4], color='deepskyblue', label=r"$\psi_4$")
ax.set_xlabel('x')
ax.set_yticks([])
plt.ylim(bottom=-0.25, top = 6)
ax.set_title(r"Numerical $\psi(x)$")
plt.legend(loc=1)

# Plot probability densities vs position
ax = fig.add_subplot(1,2,2)
ax.plot(x, x**2/2, color='silver', linestyle='--')
ax.plot(x, prob[:,0]+E[0], color='darkblue', label=r"$|\psi_0|^2$")
ax.plot(x, prob[:,1]+E[1], color='blue', label=r"$|\psi_1|^2$")
ax.plot(x, prob[:,2]+E[2], color='royalblue', label=r"$|\psi_2|^2$")
ax.plot(x, prob[:,3]+E[3], color='dodgerblue', label=r"$|\psi_3|^2$")
ax.plot(x, prob[:,4]+E[4], color='deepskyblue', label=r"$|\psi_4|^2$")
ax.set_xlabel('x')
ax.set_yticks([])
plt.ylim(bottom=-0.25, top = 5.5)
ax.set_title(r"Numerical $|\psi(x)|^2$")
plt.legend(loc=1)
plt.show()
```

To understand the accuracy of the simulation, the numerically determined eigenvalues $E_{n, \text{num}}$ are compared to the analytical values, $E_{n, \text{an}} = (n + \frac{1}{2})\hbar \omega$, using the relative error given by: $$\text{relative error in}\  E_n = \frac{|E_{n, \text{num}} - E_{n, \text{an}}|}{E_{n, \text{an}}}$$

First, consider the lowest energy states. 

```python
n_0 = np.arange(N) # integer values of n from 0 to N-1
E_analytical = (n_0+0.5)*hbar*w # analytical energies
E_error = np.abs(E - E_analytical)/ E_analytical # relative error in numerical energies

print("Numerical vs. Analytical Energies")
for i in range(5):
    print(f"n={i}:")
    print(f"  Numerical E = {E[i]:.6f}")
    print(f"  Analytical E = {E_analytical[i]:.6f}")
    print(f"  Difference = {E[i] - E_analytical[i]:.6f}")
    print(f"  Relative error in Numerical E = {E_error[i]:.6f}\n")
```

Evidently, for low energy states, the numerical eigenvalues are accurate to 3 significant figures. The error in $E_{n, \text{num}}$ is of the order $10^{-4}$. The central differences method is a second order method and for this approximation $dx = 10^{-2}$, which means $(dx)^2 = 10^{-4}$. Hence, the major source of error in the low energy states is the truncation error inherent in the numerical method used. 

To understand how the error in $E_{n, \text{num}}$ evolves as $n$ increases, consider the relative error in $E_{n, \text{num}}$ for all $n$ in the figures below.

```python
# plot relative error in E_n vs n
fig = plt.figure(figsize=(12,6))
ax1 = fig.add_subplot(1,2,1)
ax1.plot(n_0, E_error, color='forestgreen', label = 'Relative error')
ax1.axhline(0.0001, color='mediumblue', linestyle='--', label=r'Truncation error: $10^{-4}$')
#ax1.axvline(200, 0.01, 0.8)
ax1.set_xlabel(r'n')
ax1.set_ylabel(r'Relative error in $E_n$')
plt.legend()

# plot analytical and numerical energies vs n
ax2 = fig.add_subplot(1,2,2)
ax2.plot(n_0, E_analytical*10**(-3), color='k', linestyle='-.', label=r"$E_{n, \text{an}}$")
ax2.plot(n_0, E*10**(-3), color='crimson', label=r"$E_{n, \text{num}}$")
ax2.set_xlabel(r'$n$')
ax2.set_ylabel(r'$E \times 10^{-3}$')
plt.legend()
plt.show()
```

This figure indicates that for low energy eigenstates, the relative error remains on the order of $10^{-4}$, reiterating that the truncation error inherent in the central differences method is the major source of error. However, for higher energy eigenstates, a secondary source of error becomes significant and surpasses the magnitude of the truncation error, demonstrated in the left hand plot by the rapid increase in the relative error from $n \approx 200$, and in the right hand plot as $E_{n, \text{num}}$ diverges rapidly from $E_{n, \text{an}}$ for the same values of $n$. 

The following section investigates 2 possibilities for this secondary source of error; firstly the roundoff error, which, similar to the truncation error, is an inherent property of the numerical method, and secondly the error due to the domain width $L$, which arises from artificially setting $\psi = 0$ at the boundaries of the domain. ***

### 1.1) Error Analysis: Truncation vs Roundoff error

There are 2 sources of error inherent in a finite differences numerical method: roundoff errors and truncation errors. 

Roundoff errors are the difference between the exact result of a calculation and the result given by a corresponding computer algorithm. This arises as the computer tries to store infinitely-long decimal numbers with finite memory - in the case of Python, as 64-bit floating point numbers. 

For the central difference method used, the most notable cause of roundoff error is subtractive cancellation. This occurs during the subtration of two almost equal values when the cancellation of leading digits results in a value that is too small to be respresented exactly, so is set to 0. The repeated computation of $f(y_j + h) - 2f(y_j) + f(y_j - h)$ in the second derivative approximation indicates that roundoff error increases as $|h|$ decreases because this increases the loss of precision due to subtractive cancellation as the values of $f(y_j + h)$, $f(y_j)$ and $f(y_j - h)$ become closer.

Truncation error is the difference between the exact solution to an equation and the result of the approximation calculation, assuming there is no arithmetic or rounding errors. In the case of the central differences method, consider the Taylor expansion derivation for a general function $f(y)$ around point $y_j$:
$$f(y_j + h) = f(y_j) + f'(y_j)h + \frac{1}{2}f''(y_j)h^2 + \frac{1}{6}f'''(y_j)h^3 + \frac{1}{24}f^{(4)}(y_j)h^4 + \cdot$$
$$f(y_j - h) = f(y_j) - f'(y_j)h + \frac{1}{2}f''(y_j)h^2 - \frac{1}{6}f'''(y_j)h^3 + \frac{1}{24}f^{(4)}(y_j)h^4 - \cdot$$
$$f(y_j + h) + f(y_j - h) = 2f(y_j) + f''(y_j)h^2 + \frac{1}{12}f^{(4)}(y_j)h^4 + \cdots$$
$$f''(y_j) = \frac{f(y_j + h) - 2f(y_j) + f(y_j - h)}{h^2} - \frac{1}{12}f^{(4)}(y_j)h^2 - \cdots$$
Hence 
$$f''(y_j) = \frac{f(y_j + h) - 2f(y_j) + f(y_j - h)}{h^2} + \mathcal{O}(h^2) \approx \frac{f(y_j + h) - 2f(y_j) + f(y_j - h)}{h^2}$$.

Evidently, the difference between the exact expression for the second derivative and the finite approximation used is $- \frac{1}{12}f^{(4)}(y_j)h^2 - \cdots$, so this is the truncation error for this method. Since the leading error term is proportional to $h^2$, the truncation error scales quadratically with the step size and decreases as the step size decreases.

Given the relationship between the step size and both the round off and truncation errors of the method, the impact of the step size for the central differences method is investigated. This will permit an understanding of $|dx|$ and the roundoff error on the relative error in $E_{n, \text{num}}$. 

$\sin(y)$ is used as an example function (as these errors are inherent in the numerical method used, and are not unqiue to the quantum harmonic oscillator system we are considering) and the relative error at $y = \frac{\pi}{4}$ for step sizes $10^{-6} < h < 10^0$ are considered to demonstrate this relationship.. 

```python
# numerically sweep h over 6 orders of magnitude (via N and L)

L_err = 4*np.pi
N_err = np.logspace(1,7, 200)
```

```python
rel_err = np.zeros(200)

# for each value of h, find the relative error of the second derivative at y = pi/4 # 

for i in range(200):

    # sweeping h by fixing L and changing N
    N_i = int(N_err[i])
    h = L_err/N_i
    
    y_err = np.linspace(-L_err/2, L_err/2, N_i)
    s = np.sin(y_err)

    # finite differences central difference method for second derivative
    num_2der = np.zeros(N_i)
    num_2der[1:-1] = (s[2:] + s[:-2] - 2*s[1:-1])/h**2

    # k - closest y value to pi/4
    k = np.argmin(np.abs(y_err - np.pi/4))
    rel_err[i] = np.abs( - np.sin(y_err[k]) - num_2der[k])/np.sin(y_err[k]) # relative error for each dx value
```

```python
# plot relative error vs h
h = L_err/N_err

fig = plt.figure(figsize=(10,6))
ax = fig.add_subplot(1,1,1)
ax.plot(np.log10(h), np.log10(rel_err), color='forestgreen')
ax.set_xlabel(r'log$_{10}(h)$')
ax.set_ylabel(r'log$_{10}$(Relative Error at $y = \pi /4$)')
plt.show()
```

The figure shows that the relative error in the central differences method for approximating the second derivative can be minimised with respect to the step size $h$. 

For $h > h_{\text{min}}$, it is clear that the relative error increases with $|h|$, indicating that the truncation error is the primary source of error.
For $h < h_{\text{min}}$, the relative error tends to increase as $|h|$ decreases, suggesting that in this range the roundoff error becomes significant. There is considerable noise for this range of $h$ values resulting from the specific subtractive cancellation errors that occur for each step size; despite this the overall trend is clear. 

To locate the value of $h$ for which the relative error is minimised, linear regressions are completed for $h < h_{\text{min}}$ and $h > h_{\text{min}}$ using the least squares method. These fits are extrapolated until they intersect, and the value of $h$ at which they intersect is taken as $h_{\text{min}}$. 

```python
# function for straight line fits
def straight_line(x,m,c):
    y = m*x + c
    return(y)

# fit straight line parameters for h > h_min
popt, pcov = curve_fit(straight_line, np.log10(h[20:130]), np.log10(rel_err[20:130]))
m_r = popt[0] 
c_r = popt[1]

# fit straight line parameters for h < h_min
popt1, pcov1 = curve_fit(straight_line, np.log10(h[155:199]), np.log10(rel_err[155:199]))
m_l = popt1[0]
c_l = popt1[1] 
```

```python
# plot relative error vs dx with straight line fits

fig = plt.figure(figsize=(10,6))
ax = fig.add_subplot(1,1,1)
ax.plot(np.log10(h), np.log10(rel_err), color='forestgreen', label='Relative error')
ax.plot(np.log10(h), m_r*np.log10(h) + c_r, color='mediumblue', linestyle='--', label=r'Fit for $h > h_{\text{min}}$')
ax.plot(np.log10(h[130:199]), m_l*np.log10(h[130:199]) + c_l, color='magenta', linestyle=':', label=r'Fit for $h < h_{\text{min}}$')
ax.set_xlabel(r'log$_{10}(h)$')
ax.set_ylabel(r'log$_{10}$(Relative Error at $y = \pi /4$)')
plt.legend()
plt.show()
```

```python
# turning point for optimising round off vs truncation error for this numerical method

p = np.argmin(np.abs((m_r*np.log10(h) + c_r) - (m_l*np.log10(h) + c_l))) # index of dx_min
print(f"h_min = {h[p]:.9f}.")
```

To minimise the error of the central differences method for the second derivative a step size of $h = 1.5 \times 10^{-5}$ should be used. 

In the context of the quantum harmonic oscillator, a value of $dx = 10^{-5}$ would require the calculation of the eigenstates of a matrix with $10^{8} - 10^{10}$ elements. This leads this program to have an impractically long run-time. 
For a reasonable run-time, $dx = 0.01$ is used. From the figures above a step size of $0.01$ gives an error of the order $10^{-4}$, reinforcing the previous conclusion that the truncation error is the significant source of error in the lowest energy states and also indicating that at this step size value, the roundoff error is negligible compared to the truncation error. 

Therefore, we can conclude that neither the truncation or roundoff error inherent in the numerical method used can account for the rapid increase in relative error in the energies of the higher energy eigenstates of the quantum harmonic oscillator calculated. 
Consequently, the error arising as a result of the chosen domain width $L$ must be considered. 

### 1.2) Error analysis: Turning point error

The wavefunctions for the analytical eigenstates of the harmonic potential are non-zero at the boundary of the potential (where $x = \pm A_n$), indicating that there is a non-zero probability of finding a particle in a given eigenstate in the classically-forbidden region of the potential. This probability decreases exponentially with $x$ outside of the potential. 

The numerical approximation of $\hat{H}$ employed arbitrarily forces all calculated states to zero at the boundaries of the potential. To understand why this error is evident only for increasingly energetic states in figure X, we must consider the energy of a classical harmonic oscillator. 

For a classical particle with energy $E_n$, equal to that of a quantum oscillator in state $n$, at $|x| = A_n$, we have $$E_n = (n + \frac{1}{2})\hbar\omega = \frac{1}{2} m\omega^2 A_n^2$$
Hence, the classical turning point $x_{tp} = A_n$ for a particle with energy $E_n$ is $$x_{tp} = \sqrt{\frac{(2n + 1)\hbar}{m\omega}}$$

For $x_{max} >> x_{tp}$, the error arising due to the approximation $\psi_n(x_{max}) = 0$ is negligible compared to the truncation error of the numerical method, so the relative error in $E_{n, \text{num}} \approx 10^{-4}$. This is because the 'tails' of $\psi_n$ decay sufficiently between $x_{tp}$ and $x_{max}$ so that $\psi_n(x_{max}) \approx 0$. Since $x_{tp}$ increases with $n$, this indicates that this error is negligible for only the lowest energy. 

For $x_{max} << x_{tp}$, the approximation is clearly inappropriate and serves as a significant source of error in high $n$ eigenstates. 

For $x_{max} > x_{tp}$ there is some exponential decay between $x_{tp}$ and $x_{max}$ but this decay is insufficient to permit the error due to the $\psi_n(x_{max}) = 0$ approximation to be of an order less than $10^{-4}$. This causes an increase in the relative error of $E_{n, \text{num}}$ with $n$, though this is more gradual than for eigenstates where $x_{max} < x_{tp}$

The following figure demonstrates the evolution in the relative error of $E_{n, \text{num}}$ with $x_{max}$ for the first 200 eigenstates of the QHO. This is done for fixed $dx = 0.01$ so the truncation error is on the order of $10^{-4}$ for all $n$ and $x_{max}$ is varied using values of $N$ between 200 and 4000, via the relationship $x_{max} = L/2 = N dx/2$. The eigenvalues of the Hamiltonian calculated for each value of $N$ are determined and the relative error in the first 200 of these values is calculated.

```python
# vary L via N: L = N*dx
N_tp = np.linspace(200,4000, 51).astype(int) 

# exact eigenvalues for first 200 eigenstates
n = np.arange(200)
E_an = (n + 0.5)*hbar*w # analytical energy values

# stores relative error for 1st 200 eigenstates for values of N
E_rel_err = np.zeros((200,len(N_tp)))
```

```python
for j in range(len(N_tp)): # takes ~120s to run
    H_N = Hamiltonian(N_tp[j],dx)
    E_200 = eigh(H_N, eigvals_only=True, subset_by_index=[0,199]) # only calculates first 200 eigenvalues

    # calculate relative error for 1st 200 eigenstates for given N
    E_rel_err[:,j] = np.abs(E_200 - E_an)/E_an
```

```python
x_max = N_tp*dx/2

fig = plt.figure(figsize=(10,6))
ax = fig.add_subplot(1,1,1)
for b in range(10):
    ax.plot(x_max, E_rel_err[20*b+1,:], label=f'n = {20*b}')
ax.set_xlabel('x_max')
ax.set_ylabel('log(Relative Error)')
ax.set_yscale('log')
plt.legend()
plt.show()
```

The figure illustrates that for $ n \leq 80$, the relative error initially decreases as $x_{max}$ increases. By $x_{max} = 20$, the relative error plateaus on the order of $10^{-4}$, suggesting that the turning point error is less significant than the truncation error for these eigenvalues at $x_{max} = 20$.

For $80 < n < 200$, the figure shows a similar initial decrease in the relative error with increasing $x_{max}$. However, as $x_{max} \rightarrow 20$, the relative error remains on the order of $~10^{-2}$. This means that the difference between $x_{tp}$ and $x_{max}$ is insufficient to permit the eigenstates to decay enough so that the error due to the $\psi_n(x_{max}) = 0$ approximation is less significant than the truncation error. 

The increasing turning point error with $n$ for $x_{max} = 20$ suggests that this is responsible for the increase in the relative error of $E_{n, \text{num}}$ for higher energy eigenstates that is absent for lower energy eigenstates. 

# 2.0) Time Evolution of Gaussian Wavepacket in Harmonic Potential

While a quantum particle in a harmonic potential will be in a single eigenstate $\psi_n$ upon measurement at any time $t$, any one $\psi_n$ cannot model the time dependent behaviour of particle as the $\psi_n$ are stationary states, so $\langle x \rangle_n$ is time-independent. However, superposing the eigenstates forces a time-dependent $\langle x \rangle$ due to interference between the $\psi_n$, which is reflective of the observed behaviour of particles in harmonic potentials.

To model the time-dependent dynamics of a particle in the harmonic potential, a Gaussian wavepacket is constructed by superposing the numerically determined eigenstates at time $t = 0$ and propagating the calculated coefficients using the solutions to the TDSE. The normalised Gaussian wavepacket at $t = 0$ is given by: $$\Psi(x,0) = \Big( \frac{1}{2\pi\sigma^2} \Big)^{\frac{1}{4}}e^{-\frac{(x - x_0)^2}{4\sigma^2}}e^{i(k_0(x - x_0))}$$ where $\sigma$ is the initial spread, $x_0$ is the initial centre position and $k_0$ is the initial momentum of the wavepacket. 

Let $$\Psi(x,0) = \sum_n c_n \psi_n(x) $$ where $c_n$ are the complex coefficients for each eigenstate. Then consider $$\langle \psi_n | \Psi \rangle = \int \psi^*_n \Psi dx = c_n\langle \psi_n | \psi_n \rangle = c_n$$
Hence, for the discretised domain used and numerically determined eigenstates we must have $$c_n = \sum_i \psi^*_n(x_i) \Psi(x_i, 0) dx$$ where $\sum_n |c_n|^2 = 1$ for $n \in [0,N]$ since both $\Psi(x,0)$ and $\psi_n(x)$ are normalised.

The initial Gaussian wavepacket is constructed by calculating these coefficients. Given the analysis in section $(1.2)$, only the $\psi_n$ for $n$ up to $n = 80$ are used to minimise the error in the $c_n$ due to the domain width.

```python
# define parameters for intitial position, initial momentum and initial spread of wavepacket

x_0 = 0
k_0 = 10
sigma = 1
```

```python
# function for initial Gaussian wavepacket

def gaussian_wavepacket(x):
    return((1/(2*np.pi*sigma**2)**0.25)*np.exp(-((x-x_0)**2)/4*sigma**2)*np.exp(1j*k_0*(x-x_0)))
```

```python
# gaussian wavepacket at time t = 0 at each grid point

Psi_initial = np.zeros(N, dtype=complex)
for k in range(N):
    Psi_initial[k] = gaussian_wavepacket(x[k])
```

```python
# calculating coefficients c_n to decompose wavepacket into QHO eigenstates 

n_max = 80 # max number of eigenstates used - minimise error due to L

c_n = np.zeros(n_max, dtype=complex)

# calculate coefficients
for b in range(n_max):
    for a in range(N):
        c_n[b] += psi[a,b]*Psi_initial[a]*dx

Psi_sup = np.zeros(N, dtype=complex) # sum over n of c_n*\psi_n at each grid point x_i
for k in range(N):
    Psi_sup[k] = np.matmul(psi[k,:n_max],c_n)
```

```python
# plot initial Gaussian wavepacket and probability density

fig = plt.figure(figsize=(10,6)) 
ax = fig.add_subplot(1,2,1)
ax.plot(x, Psi_initial.real, color='k', linestyle='--', label='Exact Gaussian') # exact gaussian wavepacket
ax.plot(x, Psi_sup.real, color='crimson', alpha=0.7, label='Superposition') # superposition wavepacket
ax.set_ylabel(r'$\Psi(x,0)$')
ax.set_xlabel(r'$x$')
plt.legend()

ax1 = fig.add_subplot(1,2,2) 
ax1.plot(x, np.abs(Psi_initial)**2, color='k', linestyle='--', label='Exact Gaussian') # exact gaussian wavepacket
ax1.plot(x, np.abs(Psi_sup)**2, color='crimson', alpha=0.7, label=r'Superposition') # superposition wavepacket
ax1.set_ylabel(r'$|\Psi(x,0)|^2$')
ax1.set_xlabel(r'$x$')
plt.legend()

plt.show()
```

```python
print(f'The sum of |c_n|^2 for n = 0 to n = 80 is {sum(np.abs(c_n)**2):.4f}')
```

This figure demonstrates that superposing the numerically determined $\psi_n(x)$ in this way does indeed give an excellent approximation of $\Psi(x,0)$. 

By considering eigenstates for $n \leq 80$ only, we are assuming that $\sum_n ^{80} |c_n|^2 \approx \sum_n ^{N} |c_n|^2 = 1$. However, the above calculation shows that  $\sum_n ^{80} |c_n|^2 = 0.9997$, so there is a probability of $0.03\%$ not accounted for. To understand the effect of this, compare the $|c_n|^2$ calculated for eigenstates up to $\psi_{80}$ are used with those calculated using all $N$ eigenstates. 

```python
n_all_v = np.arange(N) # values of n 0 to 4000

cn_all_v = np.zeros(N, dtype=complex) # store coefficients for all eigenstates used

# calculate coefficients
for b in range(N):
    for a in range(N):
        cn_all_v[b] += psi[a,b]*Psi_initial[a]*dx

n = np.arange(n_max) # values of n 0 to 80


fig = plt.figure(figsize=(10,6))

# plot coefficients for n=0 to n=4000 eigenstates used 
ax = fig.add_subplot(1,2,1)
ax.plot(n_all_v, np.abs(cn_all_v)**2,  color='orangered') 
ax.set_xlabel(r'$n$')
ax.set_ylabel(r'$c_n^2$')
ax.set_title(r'$c_n^2$ for $n \leq N$ used')

# plot coefficients for n=0 to n=80 eigenstates used
ax1 = fig.add_subplot(1,2,2)
ax1.plot(n, np.abs(c_n)**2, color='orangered')
ax1.set_xlabel(r'$n$')
ax1.set_ylabel(r'$c_n^2$')
ax1.set_title(r'$c_n^2$ for $n \leq 80$ used')

plt.show()
```

These plots clearly show that the eigenstates that contribute significantly to the structure of the wavepacket are those between $n = 30$ and $n = 70$, and the $0.03\%$ probability not accounted for by the first $81$ states is spread thinly across the higher energy states. This assumption ($\sum_n ^{80} |c_n|^2 \approx \sum_n ^{N} |c_n|^2 = 1$) is therefore appropriate, especially when considering the turning point error that would be introduced by including the higher energy eigenstates. 

Now consider the time-evolution of the Gaussian wavepacket.

Given that the Schrodinger equation is separable for $\hat{V}(x) = \frac{1}{2} m\omega^2 x^2$, we must have that $$\psi_n(x,t) = \psi_n(x)e^{-\frac{iE_nt}{\hbar}}$$ where $e^{-\frac{iE_nt}{\hbar}}$ is a solution to the TDSE for all $n$. Hence, the Gaussian wavepacket evolves with time according to: $$\Psi(x,t) = \sum_n c_n \psi_n(x)e^{-\frac{iE_nt}{\hbar}} $$
This is implemented by calculating $\Psi(x_i,t)$ at time intervals of $0.1$ for $t \in [0,20]$ at each grid point $x_i$.

```python
# discretise time step t 
M = 201
t = np.linspace(0,20,M)

# initialise Psi_xt
Psi_xt = np.zeros((N,M), dtype=complex)

# use matrix multiplication to find time evolution from Psi_initial
for k in range(M):
    c_nk = c_n*np.exp(-1j*E[:n_max]*t[k]/hbar)
    Psi_xt[:,k] = np.matmul(psi[:, :n_max], c_nk)
```

```python
# animation showing evolution of psi^2 with time
fig = plt.figure(figsize=(10,6)) 

#set up for psi^2 
ax = plt.axes(xlim =(-20, 20), ylim =(0, 1))
ax.plot(x, np.abs(Psi_sup)**2, color='crimson', alpha=0.5, label=r'$|\Psi(x,0)|^2$')
ax.set_xlabel(r'$x$')
ax.set_ylabel(r'$|\Psi(x,t)|^2$')
plt.legend()

# plot V(x) with separate y axis
ax1 = ax.twinx()
ax1.plot(x, 0.5*m*w**2*x**2, color='gray', linestyle='--', label='V(x)')
ax1.set_ylim(0)
ax1.set_ylabel(r'$V(x)$')
plt.legend(loc = 2)

# contain psi^2
line, = ax.plot([], [], lw=2, color='crimson', label=r'$|\psi(x,t)|^2$')

def init():
    line.set_data([],[])
    return line,

# each frame is each time step in Psi_xt, 'counted' in the animation by index k 
def animate(k):
    line.set_data(x, np.abs(Psi_xt[:,k])**2)
    return line,


wavepacket_t = animation.FuncAnimation(fig, animate, init_func=init, interval=50, frames=M, blit=True)
wavepacket_t.save('wavepacket.gif', writer='pillow', fps=20)
plt.close()
```

![Wavepacket time evolution](wavepacket.gif)

From this animation, it is clear that the wavepacket does not disperse and $\Psi(x,0) = \Psi(x,mT)$ where $T = 2\pi/\omega$ is the period of oscillation and $m \in \mathbb{Z}$. This is because each contribution to the wavepacket (each $\psi_n$ with $c_n \neq 0$) accumulates phase at a rate such that the difference in phase accumulation between consecutive states is $\omega$. This means that the wavepacket is periodic and returns to its initial state with each oscillation. The linear accumulation of phase across the eigenstates occurs because the eigenvalues $E_n$ are evenly spaced - $\Delta E = \hbar\omega$ as established in section **(1.0)**. 

By including numerically determined eigenstates where $n > 80$ in the construction of the wavepacket, the non-dispersive behaviour of the wavepacket is lost. This is because the error in $E_{n, \text{num}}$ caused the spacing between energy levels to become uneven, leading consecutive eigenstates to acquire phase at rates that differ by factors not equal to $\omega$. Hence, the time-dependent behaviour of the wavepacket cannot be periodic and the wavepacket must disperse.

Also from the animation, it is evident that the spread of the wavepacket oscillates with time. To understand this behaviour, $\langle x \rangle (t)$ and $\Delta x (t)$ are considered next.

## 2.2) Position: expectation and uncertainty

The following determines the expectation values and uncertainties in the position of the Gaussian wavepacket in a harmonic potential.

For an observable $a$ with quantum operator $\hat{A}$ for any wavefunction $\psi$, the uncertainty in $a$ is given by $$\Delta a  = \sqrt{\langle a^2 \rangle - \langle a \rangle^2}$$ where $\langle a \rangle = \langle \psi | \hat{A} | \psi \rangle$ is the expectation value of $a$ and $\langle a^2 \rangle = \langle \psi | \hat{A}^2 | \psi \rangle$ is the expectation value of $a^2$.

```python
# <x> as function of t
x_exp_t = np.zeros(M)

for j in range(M):
    x_exp_t[j] = sum((np.abs(Psi_xt[:,j])**2)*x*dx)

# <x^2> as function of t
x2_exp_t = np.zeros(M)

for j in range(M):
    x2_exp_t[j] = sum((np.abs(Psi_xt[:,j])**2)*(x**2)*dx)

# uncertainty in x as function of t
x_unc_t = np.sqrt(x2_exp_t - x_exp_t**2)
```

```python
# plots for <x> and delta x respectively as functions of time
fig = plt.figure(figsize=(10,6))

ax1 = fig.add_subplot(2,1,1)
ax1.plot(t, x_exp_t, color='crimson', label='Quantum')
ax1.plot(t, 10*np.sin(t), color='lightblue',linestyle='-.', label='Classical') # x(t) = Asin(wt) is the classical solution for this potential
ax1.set_xlabel(r'$t$')
ax1.set_ylabel(r'$\langle x \rangle(t)$')
plt.legend()

ax3 = fig.add_subplot(2,1,2)
ax3.plot(t, x_unc_t, color='purple')
ax3.set_xlabel(r'$t$')
ax3.set_ylabel(r'$\Delta x(t)$')

plt.show()
```

```python
# determining the period of <x> oscillations
max_x_exp = find_peaks(x_exp_t)[0] # returns indexes of maxima in <x> excluding maxima at t = 0
T_x_exp = (t[max_x_exp[-1]]-t[max_x_exp[0]])/(len(max_x_exp)-1) # period of <x> oscillations

# determining the period of \Delta x oscillations
max_x_unc = find_peaks(x_unc_t)[0] # returns indexes of maxima in delta x excluding maxima at t = 0
T_x_unc = (t[max_x_unc[-1]]-t[max_x_unc[0]])/(len(max_x_unc)-1) # period of delta x oscillations

# determine ratio of periods
print(T_x_unc/T_x_exp)
```

The evolution of $\langle x \rangle$ with time for the Gaussian wavepacket exactly matches the trajectory of a classical particle in a harmonic potential. This is a special manifestation of Ehrenfest's theorem - which states that $\langle x \rangle$ and $\langle p \rangle$ follow approximately classical trajectories for systems with highly concentrated wavefunctions, since in this means that $\Big \langle V'(x) \Big \rangle \approx V'(\langle x \rangle)$. In the case of the quantum harmonic oscillator, the quadratic potential causes $\Big \langle V'(x) \Big \rangle = V'(\langle x \rangle)$, so $\langle x \rangle$ and $\langle p \rangle$ follow the classical trajectories exactly.  

For a standard Gaussian wavepacket, $|\psi|^2 \propto e^{-\frac{(x - x_0)^2}{2(\Delta x)^2}}$. For the wavepacket considered, at $t = 0$, $|\Psi(x,0)|^2 \propto e^{-\frac{(x - x_0)^2}{2\sigma^2}}$, which indicates that $\Delta x(0) = \sigma = 1$ as defined at the start of section **(2.0)**. The Gaussian wavepacket is a coherent state of the quantum harmonic oscillator only if $\Delta x(t) = \frac{1}{\sqrt{2}}$ for all $t$ as this minimises the product $\Delta x \Delta p$ with $\Delta x = \Delta p$. As clearly evidenced by the figure above, $\Delta x$ does not satisfy this condition, so this wavepacket cannot be a coherent state of the quantum harmonic oscillator. 

However, $\Psi(x,0)$ saturates the Heisenburg uncertainty principle: $\Delta x \Delta p = \hbar/2$. This is because a Gaussian wavefunction must achieve equality in the Cauchy-schwarz inequality. This state is non-dispersive, so $\Psi(x,t)$ remains Gaussian for all $t$.  Therefore, this wavepacket is a 'squeezed state' of the quantum harmonic oscillator - wherein the wavepacket minimises the uncertainty but, in contrast to a coherent state, the width of the wavepacket oscillates periodically with time. Since this is a squeezed state of the quantum harmonic oscillator, $\langle x \rangle(t)$ oscillates at a frequency - known as the 'breathing frequency' -  half that of $\Delta x(t)$.
